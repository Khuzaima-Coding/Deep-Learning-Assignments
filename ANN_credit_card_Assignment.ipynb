```python
#breaking our data into X and y
X=data.drop("Class",axis=1)
y=data.Class
```

```python
y.value_counts()
```

```python
#Lets we can train-test split our Data
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=123)
```

## Data Balancing

```python
#Lets we can use the SMOTE
from imblearn.over_sampling import SMOTE
smote=SMOTE(sampling_strategy="minority",random_state=0)
```

```python
X_train,y_train=smote.fit_resample(X_train,y_train)
```

```python
X_train.shape
```

```python
y_train.value_counts()
```

```python
X_train,X_val,y_train,y_val=train_test_split(X_train,y_train,test_size=0.20,stratify=y_train,random_state=42)  #stratify=y_train means that make the proportion of test(validation) ,according to the proportion of y_train. 
```

## Apply Normalization and Standardization Scaling On our Data

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
```

```python
#Normalization
mmscaler=MinMaxScaler()
mmscaled_train=mmscaler.fit_transform(X_train)
mmscaled_test=mmscaler.transform(X_test)
```

```python
#Standardization
stscaler=StandardScaler()
stscaled_train=stscaler.fit_transform(X_train)
stscaled_val=stscaler.transform(X_val)
stscaled_test=stscaler.transform(X_test)
```

# Applying Our ANN Model

```python
#Importing Libraries
from tensorflow import keras
from keras import layers
```

```python
# model_ar=keras.Sequential([
#     layers.Input(shape=(X_train.shape[1],)),  #Or (30,)
#     layers.Dense(32,activation="relu"),
#     layers.Dense(16,activation="relu"),
#     layers.Dense(1,activation="sigmoid")

# ])
```

```python
# model_ar.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy","precision","recall"])
```

```python
def create_model(activation, optimizer):
    model = keras.Sequential([
        layers.Input((X_train.shape[-1],)),
        layers.Dense(32, activation=activation),
        layers.Dense(16, activation=activation),
        layers.Dense(1, activation="sigmoid")
    ])

    model.compile(
        optimizer=optimizer,
        loss="binary_crossentropy",
        metrics=["accuracy","precision","recall"]
    )

    return model
```

```python
#model_ar.fit(stscaled_train,y_train,epochs=20,validation_split=0.20,batch_size=128)
#model_ar.fit(stscaled_train,y_train,epochs=50,validation_split=0.20,batch_size=128)
# model_ar.fit(stscaled_train,y_train,epochs=100,validation_split=0.20,batch_size=128)
```

```python
model1_50 = create_model("relu", "adam")

model1_50.fit(
    stscaled_train,
    y_train,
    epochs=50,
    validation_data=(stscaled_val,y_val),
    batch_size=128

)
```

## Model2 Fiting 

```python
model2_50 = create_model("tanh", "sgd")

model2_50.fit(
    stscaled_train,
    y_train,
    epochs=50,
    validation_data=(stscaled_val,y_val),
    batch_size=128

)
```

## Model 3 Fitting

```python
model3_100 = create_model("relu","rmsprop")

model3_100.fit(
    stscaled_train,
    y_train,
    epochs=100,
    validation_data=(stscaled_val,y_val),
    batch_size=128

)
```
