```python
data["internet_access"]=data.internet_access.map({"Yes":1,"No":0})
data["school_type"]=data.school_type.map({"Private":0,"Public":1})
data["gender"]=data.gender.map({"Male":1,"Female":0,"Other":0})
```

```python
data=pd.get_dummies(data,dtype=int)
```

```python
data.corr()
```

```python
X=data.drop(["final_grade"],axis=1)
y=data.final_grade
```

Train Test Splitting Our **Data**

```python
#Lets we can train-test split Our Data
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=0)
```

## Preprocesing

```python
#Lets we can do normalization and standardization
from sklearn.preprocessing import StandardScaler,MinMaxScaler
```

```python
stscaler=StandardScaler()
mmscaler=MinMaxScaler()
```

```python
stscaled_train=stscaler.fit_transform(X_train)
mmscaled_train=mmscaler.fit_transform(X_train)
```

```python
stscaled_test=stscaler.transform(X_test)
mmscaled_test=mmscaler.transform(X_test)
```

### Lets we can applying our ANN

```python
#Importing Libraries
from tensorflow import keras
from keras import layers
#making layers
model=keras.Sequential([
    layers.Input(shape=(X_train.shape[1],)),
    layers.Dense(128,activation="sigmoid"),
    layers.Dense(64,activation="sigmoid"),
    layers.Dense(32,activation="sigmoid"),
    layers.Dense(1,activation="linear")
])
```

```python
#model.compile(optimizer="rmsprop",loss="mean_absolute_error",metrics=["r2_score"]) #relu  #optimizer = rmsprop #score at 50epochs=0.8375,100ep=83.0
#model.compile(optimizer="sgd",loss="mean_absolute_error",metrics=["r2_score"])    #relu   #optimizer = sgd  #score at 50epochs= 0.8360,100ep=0.8381
model.compile(optimizer="adam",loss="mean_absolute_error",metrics=["r2_score"])    #relu   #optimizer = adam score at 20=0.8396,50=0.8414,100=r2_score:0.8470
```

## Fitting Our Model

```python
model.fit(stscaled_train,y_train,epochs=20,batch_size=256,validation_split=0.20)
#model.fit(stscaled_train,y_train,epochs=50,batch_size=256,validation_split=0.20)
#model.fit(stscaled_train,y_train,epochs=100,batch_size=256,validation_split=0.20)
```
